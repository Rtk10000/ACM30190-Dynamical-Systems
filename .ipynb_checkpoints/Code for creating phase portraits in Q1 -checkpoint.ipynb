```python
import numpy as np
import matplotlib.pyplot as plt
```

## Q1(i)

```python
def drdt_i(r, theta):
    dr = r**3 - 4*r
    return dr
    
def dthetadt_i(r, theta):
    dtheta = theta*0+1
    return dtheta
```

```python
r_i = np.linspace(0., 3, 40)
theta_i = np.linspace(-np.pi, np.pi, 36)

theta_s, r_s = np.meshgrid(theta_i, r_i)

dr_i = drdt_i(r_s, theta_s)
dtheta_i = dthetadt_i(r_s, theta_s)
```

```python
fig = plt.figure(figsize=(6,6))
myax = plt.axes(polar=True)

plt.streamplot(theta_i, r_i, dtheta_i, dr_i, density=0.6)

plt.ylim(0, 3)
```

## Q1(ii)

```python
def drdt_ii(r, theta):
    dr = r*(1-r**2)*(4-r**2)
    return dr
    
def dthetadt_ii(r, theta):
    dtheta = 2 - r**2
    return dtheta
```

```python
r_ii = np.linspace(0., 3, 40)
theta_ii = np.linspace(-np.pi, np.pi, 36)

theta_s, r_s = np.meshgrid(theta_ii, r_ii)

dr_ii = drdt_ii(r_s, theta_s)
dtheta_ii = dthetadt_ii(r_s, theta_s)
```

```python
fig = plt.figure(figsize=(6,6))
myax = plt.axes(polar=True)

plt.streamplot(theta_ii, r_ii, dtheta_ii, dr_ii, density=0.6)

plt.ylim(0, 3)
```

## Q1(iii)

```python
def drdt_iii(r, theta):
    dr = r*np.sin(r)
    return dr
    
def dthetadt_iii(r, theta):
    dtheta = theta*0+1
    return dtheta
```

```python
r_iii = np.linspace(0., 10, 40)
theta_iii = np.linspace(-np.pi, np.pi, 36)

theta_s, r_s = np.meshgrid(theta_iii, r_iii)

dr_iii = drdt_iii(r_s, theta_s)
dtheta_iii = dthetadt_iii(r_s, theta_s)
```

```python
fig = plt.figure(figsize=(6,6))
myax = plt.axes(polar=True)

plt.streamplot(theta_iii, r_iii, dtheta_iii, dr_iii, density=0.6)

plt.ylim(0, 10)
```

Suppose we wish to add trajectories to the plot. Below we simulate the system from Q1(ii) for a range of different initial conditions and recreate the phase portriat with these solution added in green

```python
import scipy.integrate as sci_int
```

```python
def dxdt(dt, x):
    r = x[0]
    theta = x[1]
    dr = drdt_ii(r, theta)
    dtheta = dthetadt_ii(r, theta)
    return(np.array([dr, dtheta]))
```

```python
t0=0
tn=10
ni = 101
r = np.zeros([ni,10])
th = np.zeros([ni,10]) 
```

```python
for i in range(10):
    qwe = sci_int.solve_ivp(dxdt, t_span=[t0,tn], y0=[0.21*i, 0], t_eval=np.linspace(t0,tn,ni))
    r[:,i] = qwe.y[0,:]
    th[:,i] = qwe.y[1,:]
```

```python
fig = plt.figure(figsize=(6,6))
myax = plt.axes(polar=True)

qwe = plt.plot(th, r, 'g-')
plt.streamplot(theta_ii, r_ii, dtheta_ii, dr_ii, density=0.6)

plt.ylim(0, 3)
```
