```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
import seaborn as sns
```

Define plotting style

```python
colours = sns.color_palette('deep')
```

# Phase portraits

### Competing species example

Our 2D system is defined as \begin{align}\dot{x} &= x(3-x-2y) \\ \dot{y} &= y(2-x-y).\end{align} 

```python
def f(x,y):
    return [ x * (3 - x - 2*y), y * (2 - x - y)]
```

We can use our streamplot code from before, but now we want to plot $x$ vs $y$ rather than $x$ as a funciton of $t$.

```python
# Define range of x and y values
x_range = np.linspace(0,3.5,100)
y_range = np.linspace(0,3.5,100)

# Make meshgrid
xm, ym = np.meshgrid(x_range,y_range)

# Evaluate the velocity vector for each set of x and y values
dxm, dym = f(xm,ym)

# Plot
fig = plt.figure(figsize=(6,6))
plt.streamplot(xm, ym, dxm, dym, color=colours[7])
# Add in nullclines
plt.plot(np.zeros(100),np.linspace(0,5,100),color=colours[3])
plt.plot(x_range,(3-x_range)/2,color=colours[3])

plt.plot(np.linspace(0,5,100),np.zeros(100),color=colours[0])
plt.plot(x_range,2-x_range,color=colours[0])

plt.axis([-0.05, 3.5, -0.05, 3.5])
plt.xlabel('Species 1',size= 22)
plt.ylabel('Species 2',size= 22)
plt.grid()
```

# Simulating system

Define function in correct form for using `solve_ivp`

```python
def f_ivp(t,u):
    x = u[0]
    y = u[1]
    return [ x * (3 - x - 2*y), y * (2 - x - y)]
```

Define the initial conditions and simulation length

```python
tfinal = 10
x0 = 3
y0 = 1.5
```

Simulate the system and evaluate the solution on a given time mesh

```python
sol = solve_ivp(f_ivp, [0, tfinal], [x0, y0], dense_output=True)
t = np.linspace(0, tfinal, 1000)
x,y = sol.sol(t)
```

Plot the solution as a function of time

```python
plt.figure()
plt.plot(t,x,color=colours[0],label='Species 1')
plt.plot(t,y,color=colours[2],label='Species 2')
plt.xlabel('Time',size= 20)
plt.ylabel('Population size',size= 20)
plt.legend()
```

After a question about simulating the system for a range of different initial conditions, we wrote some code to loop over range of different initial conditions and plotted the solution for each 

```python
x0_range = [0,1,2,3,4]
y0_range = [0,1,2,3,4]

plt.figure(figsize=(8,8))
# Reduce font size to avoid congestion on the figure
font = {'size': 8}
plt.rc('font', **font)

i = 1
for x0 in x0_range:
    for y0 in y0_range:
        sol = solve_ivp(f_ivp,[0, tfinal], [x0, y0], dense_output=True)
        t = np.linspace(0,tfinal,100)
        x,y = sol.sol(t)

        plt.subplot(5,5,i)
        plt.plot(t,x,color=colours[0],label='Species 1')
        plt.plot(t,y,color=colours[2],label='Species 2')
        
        plt.axis([0,10,-0.2,4.2])
        
        # only putting axis labels on outside plots, looks messy otherwise
        if i > 20:
            plt.xlabel('Time',size= 12)
        if (i+4)%5 == 0:
            plt.ylabel('Population size',size= 12)
        if i == 25:
            plt.legend()
    
        i = i+1
```

Plot solution on the phase portrait

```python
fig = plt.figure(figsize=(6,6))
# Make font big again
font = {'size': 18}
plt.rc('font', **font)

# Vector field
plt.streamplot(xm, ym, dxm, dym, color=colours[7])

# Nullclines
plt.plot(np.zeros(100),np.linspace(0,10,100), color=colours[3])
plt.plot(x_range,(3-x_range)/2, color=colours[3])
plt.plot(np.linspace(0,10,100),np.zeros(100), color=colours[0])
plt.plot(x_range,(2-x_range), color=colours[0])

# Trajectory
plt.plot(x, y, color=colours[2], linewidth=3)

plt.axis([-0.05, 3.5, -0.05, 3.5])
plt.xlabel('Rabbits',size= 28)
plt.ylabel('Sheep',size= 28)
plt.grid()
```
